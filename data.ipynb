```python
import json
import os

import numpy as np
import pandas as pd
```

```python
# folder_path = os.path.join(os.getcwd(), "runs/rss/hill instance")
folder_path = os.path.join(os.getcwd(), "runs/rss/cate estimation")
# Check if the folder exists
os.path.exists(folder_path)
```

```python
dfs = {}
folders = list(os.listdir(folder_path))

for folder in folders:
    agent_folders = list(os.listdir(os.path.join(folder_path, folder, "agents")))
    for agent_folder in agent_folders:
        with open(
            os.path.join(folder_path, folder, "agents", agent_folder, "transcript.json")
        ) as f:
            data = json.load(f)

        df = []
        for entry in data["entries"]:
            for sample in entry["samples_collection"]:
                sample_data = sample["data"]
                df.append(sample_data)

        dfs[agent_folder] = {}
        dfs[agent_folder][folder] = pd.DataFrame(df)
```

```python
for agent in dfs:
    folder_dfs = list(dfs[agent].values())
    # print(folder_dfs)
    # Check if all DataFrames are the same
    if all(df.equals(folder_dfs[0]) for df in folder_dfs):
        print(f"All DataFrames for agent {agent} are the same.")
    else:
        print(f"DataFrames for agent {agent} differ.")
```
